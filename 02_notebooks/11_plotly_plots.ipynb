```python
# import libraries
import pandas as pd
import plotly.express as px
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
names =['usman', 'ali','ahmed','umer']
heights =[5,8,6,7]
# create a dataframe
# create a line plot
fig = px.line(x=names,y=heights,title='Heights of students')
fig.show()
#  save the plot as png file
fig.write_image('./03_outputs/line_plot.png')
# save as standalone html file
fig.write_html('./03_outputs/line_plot.html')
```

```python
# creare a bar plot
px.bar(x=names,y=heights,title='Heights of students')
```

# plots on titanic dataset

```python
# import titanic dataset
df = sns.load_dataset('titanic')
df.head()
```

```python
# scatter plot between two numeric variables
fig = px.scatter(df, x='age',y= 'fare' , color = 'sex' ,title='age vs fare')
fig.show()
```

```python
df.dropna(inplace=True)
fig = px.scatter(df, x='age', y= 'fare' , color = 'sex', size='fare' ,title='plot of titanic dataset')
fig.show()
# save the plot
fig.write_image('D:/pandas_tips_tricks/02_notebooks/03_outputs.png',scale=3)
```

```python
import plotly.graph_objects as go
# create a simple line chart
fig = go.Figure(data =go.Scatter(x=[1,2,3,4],y=[10,11,12,13]))
fig.show()
```

```python
# customize the plot
fig.update_layout(
    title="Line plot of dummy data",
    xaxis_title="numbers",
    yaxis_title="values",
    showlegend=True
)
fig.show()
fig.show()
```

```python
# heatmap on correlation of titanic dataset of age and fare
fig= px.imshow(df[['age','fare']].corr(),text_auto=True)
fig.show()
```

# 15 plots in plotly


```python
df =sns.load_dataset('iris')
df.head()
```

```python
# 01 scatter plot
fig1 =px.scatter(df, x='sepal_length',y='sepal_width',color ='species')
fig1.show()
# save the plot
fig1.write_image('./03_outputs/iris_scatter.png',scale=3)
```

# 2 line plot

```python
# sort the dataset
df =df.sort_values('sepal_length')
fig2 =px.line(df, x='sepal_length',y='sepal_width',color ='species')
fig2.show()
```

# 03 bar plot

```python
# group by species and get mean based on species
# df =df.groupby('species').mean().reset_index()
# df.head()
```

```python
fig3 =px.bar(df,x='species',y='sepal_width')
fig3.show()
```

```python
# volin plot
fig4 =px.violin(df,x='species',y='sepal_width',color='species',box=True)
fig4.show()
```

```python
# histogram# histogram
fig5 = px.histogram(df, x='sepal_length', color='species')
fig5.show()
```

```python
# scatter plot in 3d
fig6 = px.scatter_3d(df, x='sepal_length',y='sepal_width',z='petal_length',color='species',size='petal_width')
fig6.show()
```
