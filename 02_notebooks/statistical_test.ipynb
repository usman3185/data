1- Chi-square test

```python
import numpy as np
from scipy.stats import chi2_contingency

# Contingency table
data = np.array([[20, 10],
                 [30, 5]])

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-square value:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
```

-The Shapiro–Wilk test is used to check whether a dataset is normally distributed

```python
import numpy as np
from scipy.stats import shapiro
```

```python
# sample data 
data = np.array([12, 15, 14, 16, 13, 15, 14, 16, 15, 14])
```

```python
stat, p_value = shapiro(data)

print("Test Statistic:", stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value > alpha:
    print("Data is normally distributed (fail to reject H0)")
else:
    print("Data is NOT normally distributed (reject H0)")
```

6️ # Important Notes

- Best for small to medium samples (n ≤ 5000)

- Very sensitive for large datasets

- Use before applying t-tests, ANOVA, regression

- In Python, a t-test is used to compare means and check whether the difference is statistically significant.

1️⃣ One-Sample t-Test

Compare the sample mean with a known population mean

```python
import numpy as np
from scipy.stats import ttest_1samp

data = np.array([12, 14, 15, 13, 16, 14, 15])
population_mean = 14

t_stat, p_value = ttest_1samp(data, population_mean)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
alpha = 0.05
if p_value > alpha:
    print("Fail to reject H₀ (means are equal)")
else:
    print("Reject H₀ (mean is significantly different)")
```

2️⃣ Independent (Two-Sample) t-Test

Compare means of two independent groups

```python
from scipy.stats import ttest_ind

group1 = np.array([12, 14, 15, 13, 16])
group2 = np.array([18, 17, 16, 19, 20])

t_stat, p_value = ttest_ind(group1, group2, equal_var=True)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
ttest_ind(group1, group2, equal_var=False)
```

3️⃣ Paired t-Test

Compare means of related samples (before vs after

```python
from scipy.stats import ttest_rel

before = np.array([50, 52, 48, 51, 49])
after  = np.array([55, 56, 54, 57, 55])

t_stat, p_value = ttest_rel(before, after)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

5️⃣ Assumptions of t-Test

✔ Data is continuous
✔ Observations are independent
✔ Data is approximately normal
✔ Equal variance (only for standard independent t-test)

The Levene’s test is used to check homogeneity of variances (i.e., whether different groups have equal variances).
It is commonly used before an independent t-test or ANOVA.

```python
import numpy as np
from scipy.stats import levene
```

```python
group1 = np.array([12, 14, 15, 13, 16])
group2 = np.array([18, 17, 16, 19, 20])
```

```python
stat, p_value = levene(group1, group2)

print("Levene Statistic:", stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value > alpha:
    print("Equal variances assumed (fail to reject H₀)")
else:
    print("Variances are NOT equal (reject H₀)")
```

```python
group3 = np.array([10, 11, 9, 12, 10])

stat, p_value = levene(group1, group2, group3)
```

```python
if p_value > 0.05:
    ttest_ind(group1, group2, equal_var=True)
else:
    ttest_ind(group1, group2, equal_var=False)  # Welch’s t-test
```

```python
import numpy as np
from scipy.stats import f_oneway
```

```python
group1 = np.array([12, 14, 15, 13, 16])
group2 = np.array([18, 17, 16, 19, 20])
group3 = np.array([10, 11, 9, 12, 10])
```

```python
f_stat, p_value = f_oneway(group1, group2, group3)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value > alpha:
    print("Fail to reject H₀ (no significant difference)")
else:
    print("Reject H₀ (significant difference exists)")
```

```python
import pandas as pd
from statsmodels.stats.multicomp import pairwise_tukeyhsd

data = np.concatenate([group1, group2, group3])
groups = (['G1']*len(group1) +
          ['G2']*len(group2) +
          ['G3']*len(group3))

df = pd.DataFrame({'value': data, 'group': groups})

tukey = pairwise_tukeyhsd(df['value'], df['group'], alpha=0.05)
print(tukey)
```

MANOVA is used when:

You have one or more categorical independent variables

AND two or more continuous dependent variables

You want to test group differences simultaneously

📌 ANOVA → one dependent variable
📌 MANOVA → multiple dependent variables

```python
import pandas as pd
import numpy as np

np.random.seed(1)

df = pd.DataFrame({
    'group': np.repeat(['A', 'B', 'C'], 20),
    'math': np.random.normal(70, 5, 60),
    'science': np.random.normal(75, 6, 60)
})
```

```python
from statsmodels.multivariate.manova import MANOVA

maov = MANOVA.from_formula('math + science ~ group', data=df)
print(maov.mv_test())
```
