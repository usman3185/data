```python
import numpy as np

# Coefficient matrix (from real & imaginary equations)
A = np.array([[2, 4], 
              [3, -1]], dtype=float)

# Constants (real & imaginary parts)
B = np.array([5, 7], dtype=float)

# Solve using numpy
solution = np.linalg.solve(A, B)

print("Solution (x, y):")
print(solution)
```

# SVD in python


```python
# generate matrix of 4 by 3 dataset
A = np.random.rand (4,3)
A
```

```python
# Lets apply SVD on this matrix
U, Sigma, VT = np.linalg.svd(A)
print("U Matrix = ", U)
print("-------------------------------------")
print("Sigma Matrix = ", Sigma)
print("-----------------------------------------")
print("VT Matrix=", VT)
```

```python
# plot the vector
import numpy as np
import matplotlib.pyplot as plt

# Define the vector
v = np.array([2, 3])

# Plot the vector starting from origin (0,0)
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1)

# Set the limits of the axes
plt.xlim(0, 5)
plt.ylim(0, 5)

# Add grid and labels
plt.grid()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Plot of Vector v = [2, 3]')

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Original vector
v = np.array([2, 3])

# Define 45 degree rotation matrix
theta = np.pi / 4  # 45 degrees in radians
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

# Rotate the vector
v_rotated = R @ v  # Matrix multiplication

print("Original vector:", v)
print("Rotated vector (45°):", v_rotated)

# --- Plot both vectors ---
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, label='Original')
plt.quiver(0, 0, v_rotated[0], v_rotated[1], angles='xy', scale_units='xy', scale=1, label='Rotated')

# Labels and limits
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Vector Rotation by 45 Degrees')
plt.axhline(0, linewidth=0.5, color='red')
plt.axvline(0, linewidth=0.5, color='black')
plt.show()
```

```python
# shear transformation matrix
import numpy as np
import matplotlib.pyplot as plt

# Original vector
v = np.array([2, 3])

# Shear transformation matrix (x-direction shear)
k = 2.5  # shear factor
S = np.array([
    [1, k],
    [0, 1]
])

# Apply shear
v_sheared = S @ v

print("Original vector:", v)
print("Sheared vector:", v_sheared)

# --- Plot both vectors with different colors ---
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='blue', label='Original')
plt.quiver(0, 0, v_sheared[0], v_sheared[1], angles='xy', scale_units='xy', scale=1, color='red', label='Sheared')

# Labels and limits
plt.xlim(-5, 20)
plt.ylim(-5, 20)
plt.grid()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Shear Transformation (X-direction)')
plt.axhline(0, linewidth=0.5, color='black')
plt.axvline(0, linewidth=0.5, color='black')

# Show legend
plt.legend()
plt.show()
```
